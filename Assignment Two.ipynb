```python
class CampusService:
    def __init__(self, service_name, location, fee):
        self.service_name = service_name
        self.location = location
        self.__fee = fee  

    def get_fee(self):
        return self.__fee

    def set_fee(self, fee):
        if fee >= 0:
            self.__fee = fee
        else:
            print("Error: Fee cannot be negative!")

    def display_service(self):
        print("Service Name:", self.service_name)
        print("Location:", self.location)
        print("Fee:", self.__fee)

class LibraryService(CampusService):
    def __init__(self, service_name, location, fee, opening_hours):
        super().__init__(service_name, location, fee)
        self.opening_hours = opening_hours

    def display_service(self):
        print("\n--- Library Service ---")
        print("Service Name:", self.service_name)
        print("Location:", self.location)
        print("Fee:", self.get_fee())
        print("Opening Hours:", self.opening_hours)

class CafeteriaService(CampusService):
    def __init__(self, service_name, location, fee, food_type):
        super().__init__(service_name, location, fee)
        self.food_type = food_type

    def display_service(self):
        print("\n--- Cafeteria Service ---")
        print("Service Name:", self.service_name)
        print("Location:", self.location)
        print("Fee:", self.get_fee())
        print("Food Type:", self.food_type)

library = LibraryService("University Library","Block A",0,"8 AM - 10 PM")
cafeteria = CafeteriaService("Campus Cafeteria","Block B",150,"Breakfast and Lunch")
cafeteria.set_fee(200)

services = [library, cafeteria]

for service in services:
    service.display_service()
```
